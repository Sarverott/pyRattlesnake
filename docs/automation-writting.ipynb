```python
# based on https://gitpython.readthedocs.io/en/stable/quickstart.html
REPO_RATTLESNAKE = git.Repo(str(
    (
        pathlib.Path(
            os.path.abspath(
                os.getcwd()
            )
        ) / ".." / ".git"
    ).resolve()
))
```

```python
# based on https://gitpython.readthedocs.io/en/stable/quickstart.html
git.Repo(str(
    (
        pathlib.Path(
            os.path.abspath(
                os.getcwd()
            )
        ) / ".." / ".git"
    ).resolve()
)).head.commit.tree
```

```python
# based on https://gitpython.readthedocs.io/en/stable/quickstart.html
REPO_RATTLESNAKE.head.commit.tree
```

```python
# based on https://gitpython.readthedocs.io/en/stable/quickstart.html
prev_commits = list(REPO_RATTLESNAKE.iter_commits(all=True, max_count=10))  # Last 10 commits from all branches.
tree = prev_commits[0].tree
```

```python
# based on https://gitpython.readthedocs.io/en/stable/quickstart.html
tree
```

```python
# based on https://gitpython.readthedocs.io/en/stable/quickstart.html
prev_commits
```

```python
# based on https://gitpython.readthedocs.io/en/stable/quickstart.html
[(entry, entry.name, entry.type) for entry in tree]
```

```python
# based on https://gitpython.readthedocs.io/en/stable/quickstart.html
def print_files_from_git(root, level=0):
    for entry in root:
        print(f"{'-' * 4 * level}| {entry.path}, {entry.type}")
        if entry.type == "tree":
            print_files_from_git(entry, level + 1)
print_files_from_git(tree)
```
